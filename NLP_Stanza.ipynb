# Demonstrating tokenization by word and sentence using Stanza, a modern NLP library developed by Stanford NLP.

```python
import stanza
nlp = stanza.Pipeline('en')
```

```python
text = """Natural language processing is a fascinating field of artificial intelligence.
It enables machines to understand and respond to human language effectively."""
doc = nlp(text)
sentences = [sentence.text for sentence in doc.sentences]
print("Stanza Sentence Tokenization:", sentences)

words = [[word.text for word in sentence.words] for sentence in doc.sentences]
print("Stanza Word Tokenization:", words)
```
