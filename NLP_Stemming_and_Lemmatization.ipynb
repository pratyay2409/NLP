# Stemming and Lemmatization

### Stemming

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.stem import PorterStemmer     # Using PorterStemmer for stemming
from nltk.tokenize import word_tokenize
```

```python
ps = PorterStemmer()   # Initialize the PorterStemmer
text ="The leaves are falling from the trees. I have been enjoying the cooler weather."
# text="He likes to fish, and he bought a new fishing rod."
words = word_tokenize(text)
stemmed_words = [ps.stem(word) for word in words]
print("Original Words:", words)
print("Stemmed Words:", stemmed_words)
```

```python
from nltk.stem import SnowballStemmer   # Using SnowballStemmer for stemming
ss = SnowballStemmer('english')  # Initialize the SnowballStemmer
text ="The leaves are falling from the trees. I have been enjoying the cooler weather."
# text="He likes to fish, and he bought a new fishing rod."
words = word_tokenize(text)
snowball_stemmed_words = [ss.stem(word) for word in words]
print("Original Words:", words)
print("Snowball Stemmed Words:", snowball_stemmed_words)
```

### Lemmatization

```python
import nltk
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger_eng')
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
from nltk.corpus import wordnet
```

```python
lemmatizer = WordNetLemmatizer()
text = "The leaves are falling from the trees. I have been enjoying the cooler weather."
tagged_words = nltk.pos_tag(word_tokenize(text))
tag_map = {'J': wordnet.ADJ, 'V': wordnet.VERB, 'N': wordnet.NOUN, 'R': wordnet.ADV}
lemmatized_words = [
    lemmatizer.lemmatize(word, tag_map.get(tag[0], wordnet.NOUN))
    for word, tag in tagged_words if word.isalpha()
]

print("Original Words:", [word for word in word_tokenize(text) if word.isalpha()])
print("Lemmatized Words:", lemmatized_words)
```
