# **Case Study-1**

### Scenario Overview:
You are acting as an e-commerce Data Analyst tasked with preparing raw customer feedback for sentiment analysis.

### Core Objective:
To preprocess customer reviews using NLTK by tokenizing the text into sentences and words, enabling text analysis at both the sentence and word levels.

### Task:

*   Tokenization & Counting
*   Sentiment Identification
*   Bonus Task (Review Flagging)

```python
import nltk
from datasets import load_dataset
from nltk.tokenize import sent_tokenize
```

```python
nltk.download('punkt',quiet=True)
nltk.download('punkt_tab',quiet=True)
```

```python
dataset = load_dataset("fancyzhx/amazon_polarity")
reviews = dataset['train']['content'][:100]
```

```python
print("=== PART 1:Sentence Tokenization & Counting ===")
all_reviews_sentences = []
for i, review in enumerate(reviews, 1):
    sentences = sent_tokenize(review)  # Tokenize into sentences
    all_reviews_sentences.append(sentences)
    num_sentences = len(sentences)     # Count the sentences
    print(f"\nReview {i}:")
    print(f"Total Sentences: {num_sentences}")
    for j, sentence in enumerate(sentences, 1):
         print(f"  [{j}] {sentence}")
```

```python
from nltk.tokenize import word_tokenize
```

```python
print("=== PART 2: Word Tokenization & Counting ===")
all_review_words = []
for i, review_sentences in enumerate(all_reviews_sentences, 1):
    print(f"\nReview {i}:")
    review_word_data = []
    for j, sentence in enumerate(review_sentences, 1):
        words = word_tokenize(sentence)      # Tokenize the sentence into words
        review_word_data.append(words)
        num_words = len(words)               # Count the words in this sentence
        clean_words_string =", ".join(words)
        print(f"  Sentence {j} ({num_words} words): {clean_words_string}")
    all_review_words.append(review_word_data)
```

```python
print("=== PART 3: Sentiment Analysis ===")
positive_lexicon = {"excellent", "satisfied", "recommend", "great", "good", "amazing","love", "best", "perfect", "fantastic", "brilliant", "outstanding","happy", "stunning", "flawless", "superb"}
negative_lexicon = {"terrible", "damaged", "bad", "poor", "worst", "awful","hate", "broken", "waste", "disappointing", "horrible","useless", "garbage", "defective", "refund", "cheap"}
for i, review_word_data in enumerate(all_review_words, 1):
    print(f"\nReview {i}:")
    found_pos_words = []
    found_neg_words = []
    for words in review_word_data:
        for word in words:
            word_lower = word.lower()
            if word_lower in positive_lexicon:
                found_pos_words.append(word_lower)
            elif word_lower in negative_lexicon:
                found_neg_words.append(word_lower)
    pos_display = ", ".join(found_pos_words) if found_pos_words else "None"
    neg_display = ", ".join(found_neg_words) if found_neg_words else "None"
    print(f"  Positive Words Count: {len(found_pos_words)} | Words: {found_pos_words}")
    print(f"  Negative Words Count: {len(found_neg_words)} | Words: {found_neg_words}")
```

```python
print("=== PART 4: Bonus Task ===")
def flag_long_review(sentence_list):
    return len(sentence_list) > 3
for i, review_sentences in enumerate(all_reviews_sentences, 1):
    num_sentences = len(review_sentences)
    if flag_long_review(review_sentences):
        print(f"Review {i}: >> FLAG: This review is long ({num_sentences} sentences). Marked for further inspection.")
    else:
        print(f"Review {i}: Normal length ({num_sentences} sentences).")
```
