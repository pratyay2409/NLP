## Word Tokenization and Sentence Tokenization(Using NLTK)

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
#Sample text for tokenizer
text="""Natural language processing (NLP) is a fascinating field of
artificial intelligence. It involves analyzing and understanding
human languages, such as English, to derive meaningful insights."""
```

```python
# Sentence Tokenization
print("Sentence Tokenization:")
sentences = sent_tokenize(text)
for i, sentence in enumerate(sentences, 1):
    print(f"Sentence {i}: {sentence}")
```

```python
# Sample text for tokenization
text="""Dr. John Smith, who earned his Ph.D. at M.I.T.,
visited NASA. He saw the Mars rover at 2:30 p.m. on 10/11/12.
Isn't it great? Contact him at dr.john@nasa.gov for more info."""
```

```python
# Sentence Tokenization
print("Sentence Tokenization:")
sentences = sent_tokenize(text)
for i, sentence in enumerate(sentences, 1):
    print(f"Sentence {i}: {sentence}")
```

```python
# word Tokenization
print("\nWord Tokenization:")
# Word Tokenization for each sentence
for i, sentence in enumerate(sentences, 1):
    words = word_tokenize(sentence)
    print(f"Words in Sentence {i}: {words}")
```

```python
# Example Telugu text
telugu_text = "నమస్కారం! ఈ రోజు మనం NLP నేర్చుకుందాం. మీరు ఎలా ఉన్నారు?"
```

```python
# Sentence Tokenization
print("Sentence Tokenization:")
sentences = sent_tokenize(telugu_text, language='english')
for i, sentence in enumerate(sentences, 1):
    print(f"Sentence {i}: {sentence}")

# Word Tokenization for each sentence
for i, sentence in enumerate(sentences, 1):
    words = word_tokenize(sentence)
    print(f"Words in Sentence {i}: {words}")
```

## Using spaCy

```python
import spacy
```

```python
nlp=spacy.load("en_core_web_sm") # Need to load the spaCy English model
```

```python
text = """Dr. John Smith, who earned his Ph.D. at M.I.T.,
visited NASA. He saw the Mars rover at 2:30 p.m. on 10/11/12.
Isn't it great? Contact him at dr.john@nasa.gov for more info."""
```

```python
doc=nlp(text)
```

```python
# Sentence Tokenization
print("--- spaCy: Sentence Tokenization ---")
for i, sentence in enumerate(doc.sents, 1):
    print(f"Sentence {i}: {sentence.text}")
```

```python
# Word Tokenization
print("\n--- spaCy: Word Tokenization ---")
for i, sentence in enumerate(doc.sents, 1):
    words = [token.text for token in sentence]
    print(f"Words in Sentence {i}: {words}")
```

## Using Stanza

```python
import stanza
```

```python
nlp=stanza.Pipeline(lang='en', processors='tokenize') #Initialize the pipeline and Indicates which model"Tokenize"
```

```python
text = """Dr. John Smith, who earned his Ph.D. at M.I.T.,
visited NASA. He saw the Mars rover at 2:30 p.m. on 10/11/12.
Isn't it great? Contact him at dr.john@nasa.gov for more info."""
```

```python
doc=nlp(text)
```

```python
# Sentence Tokenization
print("--- Stanza: Sentence Tokenization ---")
for i, sent in enumerate(doc.sentences, 1):
    print(f"Sentence {i}: {sent.text.strip()}")
```

```python
# Word Tokenization
print("\n--- Stanza: Word Tokenization ---")
for i, sentence in enumerate(doc.sentences, 1):
    words = [word.text for word in sentence.words]
    print(f"Words in Sentence {i}: {words}")
```
