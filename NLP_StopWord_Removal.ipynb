# Stopword Removal

### Demonstrates how to eliminate stopwords from a text using nltk

```python
import nltk
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
text = """Natural language processing allows computers to
understand and respond to human language in a meaningful way."""
words = word_tokenize(text)
stop_words = set(stopwords.words('english'))    # Get the list of English stopwords
filtered_words = [word for word in words if word.lower() not in stop_words]    # Remove stopwords from the tokenized words
print("Original Words:", words)
print("Filtered Words (Without Stopwords):", filtered_words)
```

### Demonstrates how to eliminate stopwords from a text using SpaCy

```python
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
text = """Natural language processing is an interesting field of artificial intelligence.
It involves understanding human languages and building models to analyze them."""
doc = nlp(text)
words = [token.text for token in doc]     # Extract all original words (tokens)
stopwords_in_corpus = [token.text for token in doc if token.is_stop]  # Get stopwords present in the text
filtered_words = [token.text for token in doc if not token.is_stop]   # Filter out stopwords from the original words
print("Original Words:", words)
print("Stopwords found in the corpus:", stopwords_in_corpus)
print("Filtered Words (Without Stopwords):", filtered_words)
```
